```python
import os
import fitz  # PyMuPDF

def extract_images_from_pdf(pdf_path, output_folder):
    # Open the PDF file
    pdf_document = fitz.open(pdf_path)

    image_count = 0

    for page_num in range(len(pdf_document)):
        page = pdf_document[page_num]
        image_list = page.get_images(full=True)

        for img_index in range(len(image_list)):
            xref = image_list[img_index][0]
            base_image = pdf_document.extract_image(xref)
            image_bytes = base_image["image"]
            image_path = os.path.join(output_folder, f'IMG_{image_count}.png')
            with open(image_path, "wb") as img_file:
                img_file.write(image_bytes)

            image_count += 1

    return image_count

# PATH TO PDF FILE
pdf_path = "F:\\LICeM\\Zeleni vodik\\Pridobivanje vodika\\Anion exchange membrane\\pdf1.pdf"



if not os.path.isfile(pdf_path):
    print("Invalid file path.")
else:
    pdf_title = os.path.splitext(os.path.basename(pdf_path))[0]
    output_folder = os.path.join(os.path.dirname(pdf_path), pdf_title)
    os.makedirs(output_folder, exist_ok=True)

    num_images_saved = extract_images_from_pdf(pdf_path, output_folder)
    print(f"{num_images_saved} images saved.")
```
