```python
import matplotlib.pyplot as plt
import numpy as np

# Style customizations for Times New Roman
font_settings = {
    "font.family": "Times New Roman",  # Closest match to the font in your image
    "font.size": 12,                  # Base font size
    "axes.titlesize": 14,             # Font size for titles
    "axes.labelsize": 12,             # Font size for axis labels
    "legend.fontsize": 10,            # Font size for the legend
    "xtick.labelsize": 10,            # Font size for x-axis ticks
    "ytick.labelsize": 10             # Font size for y-axis ticks
}

plt.rcParams.update(font_settings)  # Apply font settings globally

# Placeholder data
x = np.linspace(0, 2, 50)  # x-axis: epsilon_1
y1 = 0.4 + 0.2 * np.sin(x)  # Placeholder data for Brass(s)
y2 = 0.35 + 0.15 * np.sin(x + 0.3)  # Steel
y3 = 0.3 + 0.1 * np.sin(x + 0.6)  # Cu(s)
y4 = 0.25 + 0.05 * np.sin(x + 0.9)  # Al(s)

# Create the plot
fig, ax1 = plt.subplots(figsize=(10, 5))

# Plot the data
ax1.plot(x, y1, 'o-', label="Brass(s)", markersize=4, color='black')  # Brass(s)
ax1.plot(x, y2, 's--', label="Steel", markersize=4, color='black')    # Steel
ax1.plot(x, y3, '^-', label="Cu(s)", markersize=4, color='black')     # Cu(s)
ax1.plot(x, y4, 'd-.', label="Al(s)", markersize=4, color='black')    # Al(s)

# Labels for axes
ax1.set_xlabel(r"$\epsilon_1$, in.", fontsize=12)
ax1.set_ylabel(r"$\epsilon_2$", fontsize=12, color='black')

# Second y-axis for percentage
ax2 = ax1.twinx()
ax2.set_ylabel(r"$\epsilon_2$ (%)", fontsize=12, color='black')
ax2.set_ylim(ax1.get_ylim())  # Match limits for both y-axes

# Add a grid
ax1.grid(visible=True, which='both', linestyle='--', linewidth=0.5, color='gray')

# Add a legend
ax1.legend(title="Material", loc="upper center", bbox_to_anchor=(0.5, 1.15), ncol=4, fontsize=10, frameon=False)

# Add text block for material details
plt.text(1.8, 0.52, "Material   Maximum\nPole Height, h\nBrass(s)       2.20\nSteel             2.10\nCu(s)           1.77\nAl(s)            1.57",
         fontsize=10, va="top", ha="right", linespacing=1.5, bbox=dict(boxstyle="round", edgecolor="black", facecolor="white"))

# Adjust tick parameters
ax1.tick_params(axis='both', which='major', labelsize=10)
ax2.tick_params(axis='y', which='major', labelsize=10)

# Save the figure
plt.tight_layout()
# plt.savefig("plot_placeholder.png", dpi=300)  # Saves plot as 'plot_placeholder.png'
plt.show()
```
