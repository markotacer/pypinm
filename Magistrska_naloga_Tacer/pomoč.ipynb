```python
import numpy as np
import matplotlib.pyplot as plt
```

#### Koda za stolpec in vrstico

```python
Vektor = np.array([1,2,3])
# vrstica
vrstica = Vektor[np.newaxis]

# stolpec
stolpec = Vektor[:, np.newaxis]
print(f'Vektor shape: {Vektor.shape}')
print(f'vrstica shape: {vrstica.shape}')
print(f'stolpec shape: {stolpec.shape}')
```

Transpose ne naredi nič, tudi če izbereš prvi stolpec

```python
matrika = np.array([[1,2,3], [4,5,6]])
print(f'matrika shape: {matrika.shape}')
print(matrika[:, :])
print("----------")
print(matrika[:, 0], matrika[:, 0].shape)
print("----------")
print(matrika[:, 0].T, matrika[:, 0].T.shape)
print("----------")
print(matrika[:, [0]], matrika[:, [0]].shape)
```
