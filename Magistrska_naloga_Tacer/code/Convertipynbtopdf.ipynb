```python
import nbformat
from nbformat import v4 as nbf

# Load the original notebook
with open('MAG_01.ipynb') as f:
    nb = nbformat.read(f, as_version=4)

# Create a new notebook to hold the selected cells
new_nb = nbf.new_notebook()

# Define which cells you want to export (by index or any condition)
# For example, selecting the first 3 cells
selected_cells = nb.cells[:3]  # You can adjust this range based on the cells you want

# Add selected cells to the new notebook
new_nb.cells.extend(selected_cells)

# Save the new notebook
with open('selected_cells_notebook.ipynb', 'w') as f:
    nbformat.write(new_nb, f)

# Now you can convert the new notebook to PDF
!jupyter nbconvert --to pdf selected_cells_notebook.ipynb
```
