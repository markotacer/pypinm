**Enačbe za 3 DoF in izpisana enačba za telo 1**

$$m_1 \ddot{x}_1 
+ (c_1 + c_2) \dot{x}_1 
- c_2 \dot{x}_2 
+ (k_1 + k_2) x_1 
- k_2 x_2 
= f_1
$$


$$
\mathbf{M} =
\begin{bmatrix}
m_1 & 0   & 0   \\
0   & m_2 & 0   \\
0   & 0   & m_3
\end{bmatrix},
\quad
\mathbf{C} =
\begin{bmatrix}
c_1 + c_2 & -c_2      & 0          \\
-c_2      & c_2 + c_3 & -c_3       \\
0         & -c_3      & c_3 + c_4
\end{bmatrix},
\quad
\mathbf{K} =
\begin{bmatrix}
k_1 + k_2 & -k_2      & 0          \\
-k_2      & k_2 + k_3 & -k_3       \\
0         & -k_3      & k_3 + k_4
\end{bmatrix},

$$


$$
\mathbf{F} =
\begin{Bmatrix}
f_1 \\
f_2 \\
f_3
\end{Bmatrix},
\quad
\mathbf{x} =
\begin{Bmatrix}
x_1 \\
x_2 \\
x_3
\end{Bmatrix}
$$

$$
\mathbf{M} \ddot{\mathbf{x}} + \mathbf{C} \dot{\mathbf{x}} + \mathbf{K} \mathbf{x} = \mathbf{F}.
$$
