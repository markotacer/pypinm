```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve

# Function to calculate the endpoint of the beam given the starting point, length, and angle
def calculate_endpoint(x_start, y_start, length, angle):
    x_end = x_start + length * np.cos(angle)
    y_end = y_start + length * np.sin(angle)
    return x_end, y_end

# Define the equations
def equations(vars):
    theta1, X = vars
    theta2 = theta3 + np.pi / 2
    eq1 = xA + L[0] * np.cos(theta1) + X * np.cos(theta2) - xD - L[2] * np.cos(theta3)
    eq2 = yA + L[0] * np.sin(theta1) + X * np.sin(theta2) - yD - L[2] * np.sin(theta3)
    return [eq1, eq2]

L = np.array([1, 2, 0.25])  # Lengths of segments L1, L2, and L3
#--------------------------------------------------------
# INPUT
theta3 = 0*4.5*np.pi/18
#--------------------------------------------------------
X_initial_guess = L[1] / 2  # Initial guess for X (midpoint of [0, L[1]])
xA, yA = 0, 0
xD, yD = 0.5, 0.5
# Solve the equations
theta1, X = fsolve(equations, [0.0, X_initial_guess])
theta2 = theta3 + np.pi / 2

xB, yB = calculate_endpoint(xA, yA, L[0], theta1)
xC, yC = calculate_endpoint(xB, yB, L[1], theta2)
xE, yE= calculate_endpoint(xD, yD, L[2], theta3)
xF, yF = X*np.cos(theta2) + xB, X*np.sin(theta2) + yB

print(f'xA: {xA:.2f}, yA: {yA:.2f}')
print(f'xB: {xB:.2f}, yB: {yB:.2f}')
print(f'xC: {xC:.2f}, yC: {yC:.2f}')
print(f'xD: {xD:.2f}, yD: {yD:.2f}')
print(f'xE: {xE:.2f}, yE: {yE:.2f}')
print(f'xF: {xF:.2f}, yF: {yF:.2f}')
print('--------------------------------------------------------')
print(f'X: {X}')
print(f'Length of segment L1: {L[0]}')
print(f'Length of segment L2: {L[1]}')
print(f'Length of segment L3: {L[2]}')
print('--------------------------------------------------------')


x = np.array([xA, xB, xC, xD, xE])  # Example x coordinates
y = np.array([yA, yB, yC, yD, yE])  # Example y coordinates

x1, y1 = (xA + xB) / 2, (yA + yB) / 2
x2, y2 = (xB + xC) / 2, (yB + yC) / 2
x3, y3 = (xD + xE) / 2, (yD + yE) / 2

print(f'1: {x1:.2f},{y1:.2f}')
print(f'2: {x2:.2f},{y2:.2f}')
print(f'3: {x3:.2f},{y3:.2f}')
print(f'theta1: {theta1:.2f}')
print(f'theta2: {theta2:.2f}')
print(f'theta3: {theta3:.2f}')
print(f'theta 1: {np.rad2deg(theta1):.3f}')
print(f'theta 2: {np.rad2deg(theta2):.3f}')
print(f'theta 3: {np.rad2deg(theta3):.3f}')
print()
print(f'{x1:.5f}, {y1:.5f}, {theta1:.5f}, {x2:.5f}, {y2:.5f}, {theta2:.5f}, {x3:.5f}, {y3:.5f}, {theta3:.5f}')

# Plot lines connecting consecutive points
fig, ax = plt.subplots(figsize=(4, 3))
ax.plot([xA, xB, xC], [yA, yB, yC], 'C0o-')  # Plotting the lines
ax.plot([xD, xE], [yD, yE], 'C0o-')  # Plotting the line between D and E

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_title('Plotting Lines Connecting Consecutive Points')
ax.grid(True)
ax.axis('equal')
plt.show()

```
