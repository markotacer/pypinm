```python
import numpy as np
import matplotlib.pyplot as plt

# Define the time array
t = np.linspace(0, 0.5, num=51)  # from 0 to 0.5s with a 0.01s step

# Define the function for theta
def theta(t, theta_max, f):
    omega = 2 * np.pi * f  # Angular frequency
    return theta_max * 0.5 * (1 + np.cos(omega * t))

# Parameters
theta_max = 45  # Maximum angle in degrees
f = 10  # Frequency in Hz

# Calculate theta values
theta_values = theta(t, theta_max, f)

# Create the plot
fig, ax = plt.subplots(figsize=(5, 2))
ax.plot(t, theta_values, label='Theta')
ax.set_xlabel('Time (s)')
ax.set_ylabel('Theta (degrees)')
ax.set_title('Harmonic Variation of Theta')
ax.grid(True)
ax.legend()
plt.show()
```
