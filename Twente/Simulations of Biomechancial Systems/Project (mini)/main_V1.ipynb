## Mini project

### Basic Equations:

#### Wheel's Position and Orientation:
$$
\begin{align*}
    x(t + \Delta t) &= x(t) + \Delta t \cdot v_x(t) \\
    y(t + \Delta t) &= y(t) + \Delta t \cdot v_y(t) \\
    \theta(t + \Delta t) &= \theta(t) + \Delta t \cdot \omega(t)
\end{align*}
$$

#### Wheel's Velocity and Angular Velocity:
$$
\begin{align*}
    v_x(t + \Delta t) &= v_x(t) \\
    v_y(t + \Delta t) &= v_y(t) - g \cdot \Delta t \\
    \omega(t + \Delta t) &= \omega(t)
\end{align*}
$$

### Surface Interaction:

#### Deformation of Surface:
$$
h(t + \Delta t) = h(t) - k \cdot v_y(t) \cdot \Delta t
$$

#### Friction during Contact:
$$
F_{\text{friction}} = \mu \cdot m \cdot g
$$
