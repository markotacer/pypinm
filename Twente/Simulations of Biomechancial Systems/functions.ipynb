#### for location and second derivative

```python
import sympy as sym

# Define the symbols and functions
L1, d2, t = sym.symbols('L1 d2 t')
phi1 = sym.Function('phi1')(t)
phi2 = sym.Function('phi2')(t)
ddot_phi1 = sym.Symbol(r'\ddot{\phi}_1')  # Symbol for second derivative of phi1
ddot_phi2 = sym.Symbol(r'\ddot{\phi}_2')  # Symbol for second derivative of phi2

# Define the equation for yc2
yc2 = L1 * sym.sin(phi1) + d2 * sym.sin(phi2)

# First derivative of yc2 with respect to t
dy_c2_dt = yc2.diff(t)

# Second derivative of yc2 with respect to t
d2y_c2_dt2 = yc2.diff(t, t)

# Substitute second derivative symbols for phi1 and phi2
d2y_c2_dt2 = d2y_c2_dt2.subs(phi1.diff(t, t), ddot_phi1)
d2y_c2_dt2 = d2y_c2_dt2.subs(phi2.diff(t, t), ddot_phi2)

# Define a function to customize the multiplication symbol
def custom_mul(expr):
    return sym.latex(expr, mul_symbol='\\cdot')

# Print the equations in markdown cell format using the custom printer
print("Equation for $y_{c2}$:")
print(f"$$ {custom_mul(yc2)} $$")
print("\nFirst derivative of $y_{c2}$ with respect to t:")
print(f"$$ {custom_mul(dy_c2_dt)} $$")
print("\nSecond derivative of $y_{c2}$ with respect to t:")
print(f"$$ {custom_mul(d2y_c2_dt2)} $$")
```
