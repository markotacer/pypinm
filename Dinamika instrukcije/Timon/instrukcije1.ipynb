```python
# Parametri:
A = 10 #m
B = 4.5 #m
t1 = 5.7 #s

def funk_x(A, t):
    return A*np.sin(1*t) + A

def funk_y(B, t):
    return B*np.cos(1*t)
```

```python
x_t1= funk_x(A, t1)
y_t1= funk_y(B, t1)
# print(f"x(t1) = {x_t1:3f}m")
# print(f"y(t1) = {y_t1:3f}m")
r = np.sqrt(x_t1**2 + y_t1**2)
print(f"r = {r:3f}m")
```

```python
# Definicija simbolov
A, B, t = sp.symbols('A B t')

# Definicija funkcij
funk_x = A * sp.sin(t) + A
y = B * sp.cos(t)

# Odvod funkcij
v_x = sp.diff(funk_x, t)
v_y = sp.diff(y, t)

# Drugi odvod funkcij
a_x = sp.diff(v_x, t)
a_y = sp.diff(v_y, t)

# simbolne funkcije
print(f"Velocity in x: {v_x}")
print(f"Velocity in y: {v_y}")
print(f"Acceleration in x: {a_x}")
print(f"Acceleration in y: {a_y}")

# Vstavimo cifre (isto kot so podane v nalogi)
A_val = 10
B_val = 4.5
t_val = 5.7

v_x_1 = v_x.subs({A: A_val, t: t_val})
v_y_1 = v_y.subs({B: B_val, t: t_val})
a_x_1 = a_x.subs({A: A_val, t: t_val})
a_y_1 = a_y.subs({B: B_val, t: t_val})

print("-------------------------------------------------")
print(f"Velocity in x: {v_x_1}")
print(f"Velocity in y: {v_y_1}")
print("-------------------------------------------------")
print(f"Acceleration in x: {a_x_1}")
print(f"Acceleration in y: {a_y_1}")
print()
v = np.sqrt(float(v_x_1**2 + v_y_1**2))
a = np.sqrt(float(a_x_1**2 + a_y_1**2))

# print(type(v))
# print(type(a))

print(f"Velocity: {v:.4f}")
print(f"Acceleration: {a:.4f}")
```

```python
alpha =  np.arctan2(float(v_y_1), float(v_x_1))
alpha_deg = np.degrees(alpha)
print(f"alpha: {alpha_deg:.4f}")
print("-------------------------------------------------")

beta =  np.arctan2(float(a_y_1), float(a_x_1))
beta_deg = np.degrees(beta)
print(f"beta: {beta_deg:.4f}")
```

Tangencialni pospešej je dejansko pospešek, ki deluje na tangenti gibanja. Veva, da je hitrost odvod poti in posledično je vektor hitrosti prav tako na tangenti. Zato dejansko iščemo projekcijo pospeška na to smernico. kot med pospeško in hitrostjo je razlika v zgoraj izračunanih kotih in za projekcijo uporabimo cosinus. 

Lahko naslednjič nariševa

```python
# Tangencialni pospešek
a_t = a * np.cos(alpha - beta)
print(f"Tangencialni pospešek: {a_t:.4f}")
```

### Naloga - vrv

```python
m = 0.2
omega1 = 0.3
R = 0.65
F0 = 4.2
v1 = omega1 * R
print(f"v1 = {v1:.4f}m/s")
L1 = m * v1 * R
print(f"L1 = {L1:.4f}kgm^2/s")
```

Gibalna količina se ne ohranja, saj se obodna hitrost spremeni, vrtilna količina pa se ohranja, saj ni zinaj delujočih momentov.

Ko vrv vlečemo dol se krajša radij. Predpostavimo, da se obodna hitrost (v) ohranja, a se zaradi krajšanja radija povečuje kotna hitrost.

v = omega * r

```python
F1 = m * omega1**2 * R 
print(f"začetna centrifugalna sila F = {F1:.4f}N")

#Vrtilna kolicina se ohranja: je tudi L = m omega r**2
# pol rečeš da je L1 = L2
# dobiš da je omega2 = omega1 * (R1/R2)**2
# Pol daš to v izraz za centrifugalno silo in izraziš R2
R2 = ((m * omega1**2 * R**4) / F0)**(1/3)
print(f"R2 = {R2:.4f}m")
delta_x = R - R2
print(f"delta_x = {delta_x:.4f}m")
omega_2 = omega1 * (R/R2)**2
print(f'omega_2 = {omega_2:.5f}rad/s')
```

```python
# globals().clear()
import numpy as np
mt = 17000
ms = 40
mi = 7
v0 = 900
E_e = 250000 #J
```

```python
Ek0 = mi * v0**2 / 2
print(f"Ek0 = {Ek0:.4f}J")
p0 = mi * v0
print(f"p0 = {p0:.1f}Ns")
```

Za hitrost imamo dve enačbi in dve neznanki

$
\frac{1}{2} m_{\text{tank}} v_{\text{tank}}^2 + \frac{1}{2} m_{\text{shield}} v_{\text{shield}}^2 = 250 \, \text{kJ}
$

$
m_{\text{tank}} v_{\text{tank}} + m_{\text{shield}} v_{\text{shield}} = 0
$


```python
import sympy as sp

# Define symbols
mt, ms, E_e, vt, vs = sp.symbols('mt ms E_e vt vs')

# Equations
eq1 = sp.Eq(0.5 * mt * vt**2 + 0.5 * ms * vs**2, E_e)
eq2 = sp.Eq(mt * vt + ms * vs, 0)

# Solve equations
solutions = sp.solve((eq1, eq2), (vt, vs))

# Substitute symbols with values
mt_value = 17000  # mass of train
ms_value = 40  # mass of stone
E_e_value = 250e3  # Energy released by explosion in J

vt_value = solutions[0][0].subs({mt: mt_value, ms: ms_value, E_e: E_e_value}).evalf()
vs_value = solutions[0][1].subs({mt: mt_value, ms: ms_value, E_e: E_e_value}).evalf()
v_t = vt_value
v_s = vs_value
print("\nSubstituted values:")
print(f"v_t: {vt_value:.4f} m/s")
print(f"v_s: {vs_value:.4f} m/s")

```

```python
mt = 17000
ms = 40
mi = 7
v0 = 900
E_e = 250000 #J
v_abs = (ms*v_s - mi*v0)/(mi+ms)
print(f"v_abs = {v_abs:.4f}m/s")
print(f"v_abs = {abs(v_abs):.4f}m/s")

E_is = 0.5 * (mi+ms) * v_abs**2
print(f"E_is = {E_is:.4f}J")
```
