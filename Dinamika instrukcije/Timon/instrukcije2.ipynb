```python
r = 0.2
m1 = 3.8
m2 = 1.5
g = 9.81
```

```python
# 1: Sistem je konzervativen
# 2: Trenje ne opravlja dela, saj je kotaljenje brez podrsavanja
# 3: J_valja
r_v = 3/np.sqrt(2) * r
J_valja = m1*r_v**2
print(f'J_valja = {J_valja:.4f}')
print("-----------------")
# 4
# 5
# 6
# Ek = E_k + E_rot:  1/2 m1 v1^2 + 1/2 J omega^2 + 1/2 m2 v2^2:  
# add symbols phi, x1, x2, v1, v2, omega
import sympy as sp
phi, x1, x2, v1, v2, omega, Ek, J = sp.symbols('phi x1 x2 v1 v2 omega Ek J')
phi = 2.5 *x1
x2 = 1.5 *x1
Ek = 1/2 * m1 * v1**2 + 1/2 * J * omega**2 + 1/2 * m2 * v2**2
print(f'Ek = {Ek}')
# substitute v2 3/2 v1 and omega = 5 v1
Ek = Ek.subs(v2, 3/2 * v1)
Ek = Ek.subs(omega, 5/2 * v1)
print(f'Ek = {Ek}')
# substitute J = J_valja
Ek = Ek.subs(J, J_valja)
print(f'Ek = {Ek}')
```
