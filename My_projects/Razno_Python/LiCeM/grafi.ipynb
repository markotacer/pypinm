```python
import plotly.graph_objects as go
import numpy as np

# Define the vertices of the triangle
triangle_vertices = [(0, 0), (1, 0), (0.5, np.sqrt(3)/2)]  # Vertices of an equilateral triangle

# Define the labels for the corners
corner_labels = ['Cost', 'Productivity', 'Yield']

# Create a trace for the vertices
trace_vertices = go.Scatterternary(
    mode='markers+text',
    a=[vertex[0] for vertex in triangle_vertices],
    b=[vertex[1] for vertex in triangle_vertices],
    text=corner_labels,
    textposition='top center',
    marker=dict(size=10, color='black')
)

# Create the layout for the ternary plot
layout = go.Layout(
    ternary=dict(
        sum=1,
        aaxis=dict(title=''),
        baxis=dict(title=''),
        caxis=dict(title='')
    )
)

# Create the figure
fig = go.Figure(data=[trace_vertices], layout=layout)

# Show the figure
fig.show()
```

```python
import plotly.graph_objects as go
import numpy as np

# Define the vertices of the equilateral triangle relative to the center
triangle_vertices = [(0.5, 0), (1, np.sqrt(3)/2), (0, np.sqrt(3)/2)]

# Define the labels for the corners
corner_labels = ['Cost', 'Productivity', 'Yield']

# Create a trace for the vertices
trace_vertices = go.Scatterternary(
    mode='markers+text',
    a=[vertex[0] for vertex in triangle_vertices],
    b=[vertex[1] for vertex in triangle_vertices],
    text=corner_labels,
    textposition='top center',
    marker=dict(size=10, color='black')
)

# Create the layout for the ternary plot
layout = go.Layout(
    ternary=dict(
        sum=1,
        aaxis=dict(title=''),
        baxis=dict(title=''),
        caxis=dict(title='')
    )
)

# Create the figure
fig = go.Figure(data=[trace_vertices], layout=layout)

# Show the figure
fig.show()
```
