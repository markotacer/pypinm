```python
import sys

def convert_txt_to_csv(input_filename, output_filename):
    try:
        with open(input_filename, 'r', encoding='utf-8') as infile, \
             open(output_filename, 'w', encoding='utf-8', newline='') as outfile:
            
            for line in infile:
                # Replaces ',' with '.' and ';' with ','
                modified_line = line.replace(',', '.').replace(';', ',')
                outfile.write(modified_line)
        
        print(f"Successfully converted {input_filename} to {output_filename}")
    except FileNotFoundError:
        print(f"Error: The file {input_filename} was not found.")
    except Exception as e:
        print(f"An error occurred: {e}")

# Input and output files
List1 =  ['SAT.txt', 'SUN.txt', 'MON.txt', 'TUE.txt']
List2 =  ['SAT.csv', 'SUN.csv', 'MON.csv', 'TUE.csv']
# input_file = 'IST.txt'  # Replace this with your actual input file name
# output_file = 'IST.csv'  # Replace this with your desired output file name
for i in range(4):
    input_file = List1[i]
    output_file = List2[i]
    convert_txt_to_csv(input_file, output_file)
```
