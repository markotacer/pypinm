```python
import numpy as np
import matplotlib.pyplot as plt

# Define your function (example: log function)
def function(x):
    return np.log10(x)

# Plot points on a graph
x_values = np.array([1, 10, 100, 1000, 10000, 100000])
y_values = function(x_values)

x_values2 = np.linspace(1, 1000000, 1000000)
y_values2 = function(x_values2)

# Create the figure and axis
fig, ax = plt.subplots(figsize=(8, 6))

# Plot the data
ax.plot(x_values, y_values, 'C1o', label='log(x)')
ax.plot(x_values2, y_values2, 'C0-', label='log(x)')

# Set the title, labels, and legend
ax.set_title('Logarithmic Function')
ax.set_xlabel('x')
ax.set_ylabel('log(x)')
ax.legend(loc=(1.01, 0))

# Set the x-axis to a logarithmic scale
ax.set_xscale('log')

# Show the plot
plt.show()
```

```python
# 3. Logarithmic operations
log_a, log_b = sp.log(x), sp.log(sp.Symbol('b'))
addition = sp.simplify(log_a + log_b)  # log(a) + log(b) = log(a*b)
subtraction = sp.simplify(log_a - log_b)  # log(a) - log(b) = log(a/b)
multiplication = sp.simplify(2 * log_a)  # 2*log(a) = log(a^2)
division = sp.simplify(log_a / 2)  # log(a) / 2 = log(sqrt(a))

print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)

```

```python
# 5. Prepare specific points
points_x = [1, 2, 5]
points_y = function(np.array(points_x))
point_labels = ['A', 'B', 'C']

# Add points to plot
for i in range(len(points_x)):
    ax.scatter(points_x[i], points_y[i], color='red', zorder=3)
    ax.text(points_x[i], points_y[i], f' {point_labels[i]}', fontsize=12, verticalalignment='bottom')

# 6. Formatting
ax.set_xlabel("x")
ax.set_ylabel("log(x)")
ax.set_xlim(0, 10)
ax.set_ylim(-2, 3)
ax.legend()
ax.grid(True)

plt.show()
```
