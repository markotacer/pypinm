```python
import os
import subprocess
from PIL import Image

def pdf_to_png(input_pdf_path, output_png_path, dpi=300):
    """
    Converts a single-page PDF to a high-resolution PNG using pdftoppm.
    Args:
        input_pdf_path (str): Path to the input PDF file.
        output_png_path (str): Path to save the high-resolution PNG file.
        dpi (int): DPI for the conversion (higher means better quality).
    """
    # Ensure output path has .png extension
    if not output_png_path.lower().endswith(".png"):
        raise ValueError("Output file must have a .png extension.")

    # Use pdftoppm to convert PDF to PNG
    temp_png_path = output_png_path.replace(".png", "")  # Remove extension for pdftoppm
    command = [
        "pdftoppm",
        "-png",
        "-r", str(dpi),  # Set resolution
        input_pdf_path,
        temp_png_path  # Output without extension; pdftoppm adds "-1.png"
    ]
    try:
        subprocess.run(command, check=True)
    except FileNotFoundError:
        print("Error: Make sure pdftoppm is installed and available in your PATH.")
        return

    # Rename the output file to match the specified output_png_path
    generated_png = f"{temp_png_path}-1.png"
    os.rename(generated_png, output_png_path)
    print(f"Converted PDF to high-resolution PNG: {output_png_path}")


def remove_white_background(input_image_path, output_image_path):
    """
    Processes the image to remove the white background.
    Args:
        input_image_path (str): Path to the input image file.
        output_image_path (str): Path to save the processed image.
    """
    # Open the image and convert to RGBA
    img = Image.open(input_image_path).convert("RGBA")
    pixels = img.load()

    # Iterate over each pixel
    for y in range(img.size[1]):
        for x in range(img.size[0]):
            r, g, b, a = pixels[x, y]  # Get RGBA values
            # If the pixel is white, make it transparent
            if r > 240 and g > 240 and b > 240:  # Adjust tolerance as needed
                pixels[x, y] = (0, 0, 0, 0)  # Transparent
            else:
                pixels[x, y] = (0, 0, 0, a)  # Black with original alpha

    # Save the processed image
    img.save(output_image_path, "PNG")
    print(f"Processed image saved to {output_image_path}")


# Main program
input_pdf_path = "Trollgang.pdf"  # The PDF file in the same folder as this script
output_png_path = "Trollgang_high_res.png"  # High-resolution PNG file
processed_image_path = "Trollgang_final.png"  # Final processed image

# Ensure the input PDF exists
if not os.path.exists(input_pdf_path):
    print(f"Error: The file {input_pdf_path} does not exist in the current directory.")
else:
    # Step 1: Convert PDF to high-resolution PNG
    pdf_to_png(input_pdf_path, output_png_path)

    # Step 2: Remove the white background
    if os.path.exists(output_png_path):
        remove_white_background(output_png_path, processed_image_path)
```
