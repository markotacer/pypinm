```python
from PIL import Image, ImageDraw
import os
import io
import IPython.display

def add_image_to_canvas(image_path, canvas_size=500, top_margin=50):
    """
    Add the given image to a white canvas with specified size and margins.

    Parameters:
    image_path (str): The path to the image file.
    canvas_size (int): The size (width and height) of the square canvas.
    top_margin (int): The margin from the top of the canvas.

    Returns:
    Image object: The final image with the image centered on the canvas.
    """
    try:
        # Open the image
        img = Image.open(image_path)

        # Resize the image if needed to fit within the canvas
        max_dimension = min(canvas_size, img.width, img.height)
        img.thumbnail((max_dimension, max_dimension))

        # Create a white canvas
        canvas = Image.new("RGB", (canvas_size, canvas_size), "white")

        # Calculate position to center the image
        x_offset = (canvas_size - img.width) // 2
        y_offset = (canvas_size - img.height) // 2 + top_margin

        # Paste the image onto the canvas
        canvas.paste(img, (x_offset, y_offset))

        # Display preview of the final image
        display(canvas)

        return canvas
    except FileNotFoundError:
        print("File not found. Please check the file path.")
        return None
    except Exception as e:
        print(f"An error occurred: {e}")
        return None
```

```python
from PIL import Image, ImageDraw
import os
import io
import IPython.display

def add_image_to_canvas(image_path, canvas_size=(500, 500), margin=50):
    """
    Add the given image to a white canvas with specified size and margin.

    Parameters:
    image_path (str): The path to the image file.
    canvas_size (tuple): The size (width and height) of the square canvas. Default is (500, 500).
    margin (int): The margin around the image. Default is 50 pixels.

    Returns:
    Image object: The final image with the image centered on the canvas.
    """
    try:
        # Open the image
        img = Image.open(image_path)

        # Calculate max dimension for resizing
        max_dimension = min(canvas_size)

        # Resize the image if needed to fit within the canvas
        img.thumbnail((max_dimension, max_dimension))

        # Calculate margins for each side
        top_margin = (canvas_size[1] - img.height) // 2
        bottom_margin = canvas_size[1] - img.height - top_margin
        left_margin = (canvas_size[0] - img.width) // 2
        right_margin = canvas_size[0] - img.width - left_margin

        # Create a white canvas with margins
        canvas = Image.new("RGB", (canvas_size[0], canvas_size[1]), "white")

        # Paste the image onto the canvas with margins
        canvas.paste(img, (left_margin, top_margin))

        # Display preview of the final image
        display(canvas)

        return canvas
    except FileNotFoundError:
        print("File not found. Please check the file path.")
        return None
    except Exception as e:
        print(f"An error occurred: {e}")
        return None

# Example usage
# Define the parameters
image_path = "G:/07_SLIKE/2024/01_Room_tours/IMG_2021.jpg"
canvas_size = (800, 800)
margin = 100

# Check if the path exists
if os.path.exists(image_path):
    # Add the image to the canvas
    final_image = add_image_to_canvas(image_path, canvas_size, margin)
    if final_image:
        # Output format
        output_format = input("Enter the output format (e.g., JPEG, PNG): ").upper()

        # Save the final image
        output_path = "final_image." + output_format.lower()
        final_image.save(output_path)
        print(f"Final image saved as {output_path}")
else:
    print("Invalid path. Please enter a valid path to the image.")
```
