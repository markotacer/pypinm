```python
# Import the required library
from PIL import Image
from IPython.display import display

def remove_white_background(input_image_path, output_image_path):
    """
    Processes the image to remove the white background.
    Args:
        input_image_path (str): Path to the input image file.
        output_image_path (str): Path to save the processed image.
    """
    # Open the image and convert to RGBA
    img = Image.open(input_image_path).convert("RGBA")
    pixels = img.load()

    # Iterate over each pixel
    for y in range(img.size[1]):
        for x in range(img.size[0]):
            r, g, b, a = pixels[x, y]  # Get RGBA values
            # If the pixel is white, make it transparent
            if r > 240 and g > 240 and b > 240:  # Adjust tolerance as needed
                pixels[x, y] = (0, 0, 0, 0)  # Transparent
            else:
                pixels[x, y] = (0, 0, 0, a)  # Black with original alpha

    # Save and display the new image
    img.save(output_image_path, "PNG")
    print(f"Processed image saved to {output_image_path}")
    display(img)  # Display the processed image in the notebook

# Example usage
input_path = "Trollgang.pdf"  # Replace with your input image path
output_path = "TG.png"  # Replace with your output image path
remove_white_background(input_path, output_path)

```

```python
from PIL import Image, ImageEnhance, ImageFilter
def enhance_image_quality(image_path="stš_out.png", output_image_path="stš_enhanced.png"):
    """
    Enhances the quality of the image by sharpening and resizing.
    Args:
        image_path (str): Path to the processed image file.
        output_image_path (str): Path to save the enhanced image.
    """
    # Open the processed image
    img = Image.open(image_path)
    
    # Apply sharpening filter
    img = img.filter(ImageFilter.SHARPEN)
    
    # Enhance contrast and brightness
    enhancer = ImageEnhance.Contrast(img)
    img = enhancer.enhance(1.5)  # Increase contrast (adjust value as needed)
    
    enhancer = ImageEnhance.Brightness(img)
    img = enhancer.enhance(1.2)  # Increase brightness (adjust value as needed)
    
    # Optionally resize to higher resolution
    img = img.resize((img.size[0] * 2, img.size[1] * 2), Image.Resampling.LANCZOS)
    
    # Save and display the enhanced image
    img.save(output_image_path, "PNG")
    print(f"Enhanced image saved to {output_image_path}")
    display(img)  # Display the enhanced image in the notebook

# Example usage for quality enhancement
enhance_image_quality()


```
